```python
%load_ext autoreload
%autoreload 2

import os
import x4c  # https://ncar.github.io/x4c/
import xarray as xr
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
print(x4c.__version__)
os.chdir('/glade/work/fengzhu/Projects/paper-MCO_iCESM/notebooks')
```

## Load Inference Result

```python
df_CO2pdf = pd.read_csv('../data/df_CO2pdf.csv', index_col=0)
idx = np.argmax(df_CO2pdf['density'])
print(idx)
print(df_CO2pdf.loc[idx])
```

```python
# load useful variables
dz = x4c.open_dataset('../data/iCESM1.3_dz.nc')['dz']
TAREA = x4c.open_dataset('../data/iCESM1.3_TAREA.nc')['TAREA']
```

## TEMP

```python
vn = 'TEMP'
case = {}
case['3x'] = xr.open_dataarray(f'../data/iCESM1.3_3x_{vn}_clim_eq.nc')
case['1.5x'] = xr.open_dataarray(f'../data/iCESM1.3_1.5x_{vn}_clim_eq.nc')
case['6x'] = case['3x'] + (case['3x'] - case['1.5x'])
interp = np.linspace(case['1.5x'].values, case['6x'].values, len(df_CO2pdf))
case['mle'] = case['3x'].copy()
case['mle'].values = interp[idx]

ds = case['mle'].to_dataset()
ds['dz'] = dz
ds['gw'] = TAREA
ds.attrs['corresponding_author'] = 'Feng Zhu (fengzhu@ucar.edu)'
ds.attrs['citation'] = 'Zhu, F., and J. Zhu. 2025. Long Simulations of the Miocene Climatic Optimum. NSF National Center for Atmospheric Research. https://doi.org/10.5065/3QFN-GN70.'
ds.to_netcdf(f'../data/iCESM1.3_MLE_{vn}_clim_eq.nc')
```

## R18O climo

```python
vn = 'd18Osw'
case = {}
case['3x'] = xr.open_dataarray(f'../data/iCESM1.3_3x_{vn}_climo.nc').isel(z_t=0) - 0.55 # MCO global volume mean assumed
case['1.5x'] = xr.open_dataarray(f'../data/iCESM1.3_1.5x_{vn}_climo.nc').isel(z_t=0) - 0.55 # MCO global volume mean assumed
case['6x'] = case['3x'] + (case['3x'] - case['1.5x'])
interp = np.linspace(case['1.5x'].values, case['6x'].values, len(df_CO2pdf))
case['mle'] = case['3x'].copy()
case['mle'].values = interp[idx]
da = case['mle']/1e3 + 1
da.name = 'R18O'

ds = da.to_dataset()
ds['gw'] = TAREA
ds.attrs['corresponding_author'] = 'Feng Zhu (fengzhu@ucar.edu)'
ds.attrs['note'] = 'The global volume mean of d18Osw is set to -0.55 permil.'
ds.attrs['citation'] = 'Zhu, F., and J. Zhu. 2025. Long Simulations of the Miocene Climatic Optimum. NSF National Center for Atmospheric Research. https://doi.org/10.5065/3QFN-GN70.'
ds.to_netcdf(f'../data/iCESM1.3_MLE_R18O_climo.nc')
ds
```
